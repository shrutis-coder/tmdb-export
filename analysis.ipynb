```python
import matplotlib.pyplot as plt

# Sample DataFrame creation (you can skip this if you already have df)
# import pandas as pd
# df = pd.DataFrame({
#     'runtime': [90, 120, 150, 180, 200],
#     'vote_average': [5.5, 6.7, 8.0, 7.5, 9.0]
# })

# Create a scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['runtime_minutes'], df['vote_average'], alpha=0.7, color='blue')

# Adding titles and labels
plt.title('Scatter Plot of Runtime vs Vote Average')
plt.xlabel('Runtime (minutes)')
plt.ylabel('Vote Average')

# Show grid
plt.grid()

# Show the plot
plt.show()
```

```python
import pandas as pd

# Calculate ROI and add it as a new column
df['ROI'] = (df['revenue'] - df['budget']) / df['budget']

# Display the updated DataFrame
print(df)
```

```python
# Assuming 'df' is your DataFrame and it has 'production_countries' and 'ROI' columns
average_roi = df.groupby('production_countries')['ROI'].mean().reset_index()

# Display the result
average_roi
```

```python
import pandas as pd

import matplotlib.pyplot as plt

# Assuming df is your existing DataFrame and 'roi_column' is the name of the ROI column
# Group by 'production_countries' and calculate the average ROI
average_roi = df.groupby('production_countries')['ROI'].mean().reset_index()

# Sort the values for better visualization
average_roi = average_roi.sort_values(by='ROI', ascending=False)

# Plotting
plt.figure(figsize=(12, 6))
plt.bar(average_roi['production_countries'], average_roi['ROI'], color='skyblue')
plt.xlabel('Production Countries')
plt.ylabel('Average ROI')
plt.title('Average ROI by Production Countries')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
